```python
# Q6: Create and save the employee dataset
import pandas as pd

employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees_df = pd.DataFrame(employee_data)
employees_df.to_csv("employees.csv", index=False)
display(employees_df)
```

```python
# Q6(a-b): Inspect the DataFrame
print("Shape:", employees_df.shape)
print("Summary:")
employees_df.info()
```

```python
# Q6(c-d): Descriptive statistics and selected rows
print("Descriptive statistics:")
display(employees_df.describe())

print("First five rows:")
display(employees_df.head())

print("Last three rows:")
display(employees_df.tail(3))
```

```python
# Q6(e): Calculate the requested statistics
print("Average salary:", employees_df["Salary"].mean())
print("Total bonus:", employees_df["Bonus"].sum())
print("Youngest employee age:", employees_df["Age"].min())
print("Highest performance rating:", employees_df["Rating"].max())
```

```python
# Q6(f): Sort employees by salary in descending order
employees_sorted = employees_df.sort_values(by="Salary", ascending=False)
display(employees_sorted)
```

```python
# Q6(g): Categorize employees by performance rating
employees_df["Performance_Category"] = "Average"
employees_df.loc[employees_df["Rating"] >= 4.5, "Performance_Category"] = "Excellent"
employees_df.loc[(employees_df["Rating"] >= 4.0) & (employees_df["Rating"] < 4.5), "Performance_Category"] = "Good"
display(employees_df[["Name", "Rating", "Performance_Category"]])
```

```python
# Q6(h-j): Missing values, rename, and employee filters
print("Missing values:")
display(employees_df.isnull().sum())

employees_df = employees_df.rename(columns={"Employee_ID": "ID"})
print("Employees with more than 5 years of experience:")
display(employees_df[employees_df["Years_of_Experience"] > 5])

print("Employees in the IT department:")
display(employees_df[employees_df["Department"] == "IT"])
```

```python
# Q6(k-l): Add tax and save the modified DataFrame
employees_df["Tax"] = employees_df["Salary"] * 0.10
employees_df.to_csv("employees_modified.csv", index=False)
print("Modified dataset saved as employees_modified.csv")
display(employees_df)
```
