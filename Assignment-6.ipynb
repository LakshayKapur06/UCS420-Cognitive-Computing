### Q1. Consider the following sensor readings:
```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
```
A sensor has an average measurement error of +2°C.

Perform the following without using a for loop:
a. Add 2°C to every reading using vectorization.
b. Convert all temperatures from Celsius to Fahrenheit using:
   $$F = \frac{9}{5}C + 32$$
c. Identify all readings greater than 32°C using Boolean indexing.
d. Count how many readings exceed 32°C.
e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value.


````python
import numpy as np

# Sensor temperature readings
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original temperature readings:", temperature)

# a) Add 2 C to every reading using vectorization
corrected_temperature = temperature + 2
print("a) Corrected readings (+2 C):", corrected_temperature)

# b) Convert Celsius to Fahrenheit
fahrenheit = (9 / 5) * corrected_temperature + 32
print("b) Temperature in Fahrenheit:", fahrenheit)

# c) Identify readings greater than 32 C using Boolean indexing
readings_greater_32 = corrected_temperature[corrected_temperature > 32]
print("c) Readings greater than 32 C:", readings_greater_32)

# d) Count how many readings exceed 32 C
count_greater_32 = np.sum(corrected_temperature > 32)
print("d) Count exceeding 32 C:", count_greater_32)

# e) Explanation
print("""
e) Explanation:
Vectorization replaces explicit Python-level loops with optimized C-level SIMD operations,
drastically reducing interpreter overhead and accelerating processing. Boolean indexing constructs
a vectorized boolean mask in memory and filters elements in a single contiguous pass, making
data manipulation more expressive, concise, and memory-efficient.
""")
````

### Q2. The following matrix represents the number of daily steps recorded by four users over three days:
```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
```
Perform the following:
a. Calculate the total steps recorded.
b. Calculate the mean number of steps.
c. Find the maximum and minimum values.
d. Calculate the total steps for each day using axis=0.
e. Calculate the total steps for each user using axis=1.
f. Find the user/day position containing the maximum number of steps using argmax().


````python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps matrix (4 users x 3 days):\n", steps)

# a) Total steps recorded
total_steps = np.sum(steps)
print(f"\na) Total steps recorded: {total_steps}")

# b) Mean number of steps
mean_steps = np.mean(steps)
print(f"b) Mean steps: {mean_steps:.2f}")

# c) Maximum and minimum values
print(f"c) Maximum steps: {np.max(steps)} | Minimum steps: {np.min(steps)}")

# d) Total steps for each day (axis=0)
print(f"d) Total steps for each day (axis=0): {np.sum(steps, axis=0)}")

# e) Total steps for each user (axis=1)
print(f"e) Total steps for each user (axis=1): {np.sum(steps, axis=1)}")

# f) User/day position containing maximum number of steps
max_pos = np.unravel_index(np.argmax(steps), steps.shape)
print(f"f) Maximum steps position: User {max_pos[0] + 1}, Day {max_pos[1] + 1} (value = {steps[max_pos]})")
````

### Q3. Write a Python program using NumPy to perform the following operations:
a. Create the following NumPy array:
   `original = np.array([1, 2, 3, 4, 5, 6])`
b. Create a slice containing elements from index 1 to 4 and store it in a variable named subset.
c. Modify the first element of subset to 999 and display both original and subset.
d. Create another slice from original, but this time use .copy(). Modify its first element to 500 and display original and the copied array.


````python
import numpy as np

# a) Create original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original array:", original)

# b) Create slice from index 1 to 4
subset = original[1:4]
print("Subset slice [1:4]:", subset)

# c) Modify first element of subset to 999
subset[0] = 999
print("\nAfter modifying subset[0] to 999:")
print("Original:", original)
print("Subset  :", subset)
print("-> Note: The original array is modified because basic slicing produces a view.")

# d) Create another slice using .copy()
copied_array = original[1:4].copy()
copied_array[0] = 500
print("\nAfter modifying copied_array[0] to 500:")
print("Original    :", original)
print("Copied array:", copied_array)
print("-> Note: The original array remains unchanged because .copy() produces an independent copy.")
````

### Q4. A simple cognitive-assistive system wants to predict a user's assistance score based on three factors:
• Sleep hours
• Activity level
• Stress level

Consider:
```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
```
Here, each row represents one user and each column represents a feature.

Perform the following:
a. Display the shape and dimensions of X.
b. Calculate: X.T and explain what the transpose represents.
c. Calculate the matrix product: X.T @ X
d. Calculate the matrix inverse using: np.linalg.inv() where appropriate.
e. Implement the Ordinary Least Squares equation: $\beta = (X^T X)^{-1} X^T y$ using NumPy.
f. Explain what the resulting coefficients represent in the context of the three user features.
g. A new user has: new_user = np.array([5, 40, 7]). Use your calculated coefficients to obtain a predicted assistance score.


````python
import numpy as np

# Feature matrix X and target y
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

# a) Shape and dimensions of X
print(f"a) Shape of X: {X.shape} | Dimensions of X: {X.ndim}")

# b) Calculate X.T and explain
XT = X.T
print("\nb) X.T (Transpose of X):\n", XT)
print("""Explanation: In X, rows represent users and columns represent features (sleep, activity, stress).
In X.T, rows correspond to features and columns correspond to individual users.
""")

# c) Calculate X.T @ X
XTX = XT @ X
print("c) Matrix product X.T @ X:\n", XTX)

# d) Matrix inverse of X.T @ X
XTX_inv = np.linalg.inv(XTX)
print("\nd) Inverse of X.T @ X:\n", XTX_inv)

# e) OLS Equation: beta = (X.T @ X)^(-1) @ X.T @ y
beta = XTX_inv @ XT @ y
print("\ne) OLS Coefficients (beta):", beta)

# f) Explain coefficients
print("""
f) Coefficient Interpretation:
   beta[0] (Sleep Hours)   = {:.4f}
   beta[1] (Activity Level) = {:.4f}
   beta[2] (Stress Level)   = {:.4f}
Each coefficient indicates the marginal change in the predicted cognitive assistance score for a one-unit
increase in that respective feature, holding all other features constant.
""".format(beta[0], beta[1], beta[2]))

# g) Predict assistance score for new user: [5, 40, 7]
new_user = np.array([5, 40, 7])
pred_score = new_user @ beta
print(f"g) Predicted assistance score for new user [5, 40, 7]: {pred_score:.2f}")
````
