```python
import numpy as np

# Sensor temperature readings
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add 2°C to every reading using vectorization
corrected_temperature = temperature + 2

print("Original temperature readings:")
print(temperature)

print("\nCorrected temperature readings:")
print(corrected_temperature)


# b) Convert Celsius to Fahrenheit
# Formula: F = (9/5)C + 32
fahrenheit = (9 / 5) * corrected_temperature + 32

print("\nTemperature in Fahrenheit:")
print(fahrenheit)


# c) Identify all readings greater than 32°C using Boolean indexing
readings_greater_32 = corrected_temperature[corrected_temperature > 32]

print("\nReadings greater than 32°C:")
print(readings_greater_32)


# d) Count how many readings exceed 32°C
count_greater_32 = np.sum(corrected_temperature > 32)

print("\nNumber of readings greater than 32°C:")
print(count_greater_32)


# e) Explanation
print("\nExplanation:")
print("Vectorization allows NumPy to perform an operation on the entire array")
print("at once instead of explicitly processing each element using a for loop.")
print("Boolean indexing creates a Boolean condition for every element and")
print("directly selects the elements that satisfy the condition.")
print("This makes numerical data processing shorter, faster, and more efficient.")
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps matrix:")
print(steps)


# a) Calculate total steps recorded
total_steps = np.sum(steps)

print("\na) Total steps:", total_steps)


# b) Calculate mean number of steps
mean_steps = np.mean(steps)

print("b) Mean number of steps:", mean_steps)


# c) Find maximum and minimum values
maximum_steps = np.max(steps)
minimum_steps = np.min(steps)

print("c) Maximum steps:", maximum_steps)
print("   Minimum steps:", minimum_steps)


# d) Total steps for each day using axis=0
total_each_day = np.sum(steps, axis=0)

print("d) Total steps for each day:", total_each_day)


# e) Total steps for each user using axis=1
total_each_user = np.sum(steps, axis=1)

print("e) Total steps for each user:", total_each_user)


# f) Find user/day position containing maximum number of steps
max_position = np.unravel_index(np.argmax(steps), steps.shape)

print("f) Position of maximum number of steps:", max_position)
print("   Maximum value:", steps[max_position])
print("   User number:", max_position[0] + 1)
print("   Day number:", max_position[1] + 1)
```

```python
import numpy as np

# a) Create the original array
original = np.array([1, 2, 3, 4, 5, 6])

print("Original array:")
print(original)


# b) Create a slice from index 1 to 4
# This includes indices 1, 2, and 3
subset = original[1:4]

print("\nSubset:")
print(subset)


# c) Modify the first element of subset to 999
subset[0] = 999

print("\nAfter modifying subset:")
print("Original:", original)
print("Subset:", subset)

print("\nThe original array is also affected because a normal NumPy slice")
print("creates a view of the original array rather than an independent copy.")


# d) Create another slice using .copy()
copied_array = original[1:4].copy()

# Modify its first element
copied_array[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

print("\nThe original array is not affected because .copy() creates")
print("an independent copy of the data.")
```

```python
import numpy as np

# Create the 3 x 4 matrix again
matrix = np.arange(1, 13).reshape(3, 4)

# g) Flatten using flatten() and ravel()
flattened_array = matrix.flatten()
ravel_array = matrix.ravel()

print("Original matrix:")
print(matrix)

print("\nUsing flatten():")
print(flattened_array)

print("\nUsing ravel():")
print(ravel_array)


# h) Modify an element in ravel()
ravel_array[0] = 999

print("\nAfter modifying ravel()[0] to 999:")
print("Ravel array:")
print(ravel_array)

print("Original matrix:")
print(matrix)

print("\nThe original matrix is affected because ravel() normally")
print("returns a view of the original array when possible.")


# i) Modify an element in flatten()
flattened_array[1] = 500

print("\nAfter modifying flatten()[1] to 500:")
print("Flattened array:")
print(flattened_array)

print("Original matrix:")
print(matrix)

print("\nThe original matrix is NOT affected because flatten()")
print("returns a copy of the array.")
```

```python
import numpy as np

# Feature matrix
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

# Target values
y = np.array([40, 65, 30, 85])


# Calculate X.T @ X
XTX = X.T @ X

print("X.T @ X:")
print(XTX)


# d) Calculate inverse of X.T @ X
XTX_inverse = np.linalg.inv(XTX)

print("\nInverse of X.T @ X:")
print(XTX_inverse)


# e) Ordinary Least Squares equation:
# beta = (X.T @ X)^(-1) @ X.T @ y

beta = XTX_inverse @ X.T @ y

print("\nOLS coefficients (beta):")
print(beta)


# f) Explain coefficients
print("\nExplanation of coefficients:")
print("Coefficient 1 corresponds to Sleep Hours.")
print("Coefficient 2 corresponds to Activity Level.")
print("Coefficient 3 corresponds to Stress Level.")
print("Each coefficient represents the estimated change in the")
print("assistance score associated with a one-unit change in that")
print("feature, while the other features are held constant.")


# g) Predict assistance score for new user
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("\nNew user:")
print(new_user)

print("\nPredicted assistance score:")
print(predicted_score)
```
