### Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:
a) Addition of 2 in all the element
b) Multiply 3 with all the elements
c) Divide every element by 2


```python
import numpy as np

# Create a 1-D NumPy array with 5 elements
arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)

# a) Addition of 2 to all elements
print("a) After adding 2:", arr + 2)

# b) Multiply 3 with all elements
print("b) After multiplying by 3:", arr * 3)

# c) Divide every element by 2
print("c) After dividing by 2:", arr / 2)
```

### Q.2 (b) Find the most frequent value and their indice(s) in the following arrays:
i. x = np.array([1,2,3,4,5,1,2,1,1,1])
ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])


```python
import numpy as np

# Array x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values_x, counts_x = np.unique(x, return_counts=True)
max_count_x = counts_x.max()
most_frequent_x = values_x[counts_x == max_count_x]

print("Array x:", x)
print("Most frequent value(s):", most_frequent_x)
print("Maximum frequency:", max_count_x)
for value in most_frequent_x:
    indices = np.where(x == value)[0]
    print(f"Value {value} occurs at indices: {indices}")

# Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values_y, counts_y = np.unique(y, return_counts=True)
max_count_y = counts_y.max()
most_frequent_y = values_y[counts_y == max_count_y]

print("\nArray y:", y)
print("Most frequent value(s):", most_frequent_y)
print("Maximum frequency:", max_count_y)
for value in most_frequent_y:
    indices = np.where(y == value)[0]
    print(f"Value {value} occurs at indices: {indices}")
```

### Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?


```python
import numpy as np

# Create 1-D array named Lakshay
Lakshay = np.linspace(10, 100, 25)

print("Array Lakshay:")
print(Lakshay)

# Dimensions, Shape, Size, Dtype, Nbytes
print(f"\nDimensions (ndim)       : {Lakshay.ndim}")
print(f"Shape                   : {Lakshay.shape}")
print(f"Total elements (size)   : {Lakshay.size}")
print(f"Data type (dtype)       : {Lakshay.dtype}")
print(f"Total bytes consumed    : {Lakshay.nbytes} bytes")

# Transpose using reshape()
Lakshay_transpose = Lakshay.reshape(25, 1)
print(f"\nTranspose using reshape(25, 1): shape = {Lakshay_transpose.shape}")

# Transpose using T attribute
Lakshay_T = Lakshay.T
print(f"Transpose using .T: shape = {Lakshay_T.shape}")

print("""
Can we do the same with T attribute?
Explanation: For a 1-D NumPy array, the .T attribute has no effect because 1-D arrays have only one axis.
To represent a transpose of a vector as a column vector, reshape(25, 1) or np.newaxis is required.
""")
```

### Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.


```python
import numpy as np

# Create 3 x 4 array
ucs420_Lakshay = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array ucs420_Lakshay (3 x 4):")
print(ucs420_Lakshay)

# Statistical computations
print(f"\nMean           : {np.mean(ucs420_Lakshay)}")
print(f"Median         : {np.median(ucs420_Lakshay)}")
print(f"Maximum        : {np.max(ucs420_Lakshay)}")
print(f"Minimum        : {np.min(ucs420_Lakshay)}")
print(f"Unique elements: {np.unique(ucs420_Lakshay)}")

# Reshape to four rows and three columns
reshaped_ucs420_Lakshay = ucs420_Lakshay.reshape(4, 3)
print("\nreshaped_ucs420_Lakshay (4 x 3):")
print(reshaped_ucs420_Lakshay)

# Resize to two rows and three columns
resized_ucs420_Lakshay = np.resize(ucs420_Lakshay, (2, 3))
print("\nresized_ucs420_Lakshay (2 x 3):")
print(resized_ucs420_Lakshay)
```
