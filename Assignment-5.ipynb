```python
import numpy as np

# Create a 1-D NumPy array with 5 elements
arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)

# a) Addition of 2 to all elements
print("After adding 2:", arr + 2)

# b) Multiply 3 with all elements
print("After multiplying by 3:", arr * 3)

# c) Divide every element by 2
print("After dividing by 2:", arr / 2)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

# Reverse the array
reversed_arr = arr[::-1]

print("Original array:", arr)
print("Reversed array:", reversed_arr)
```

```python
import numpy as np

# Array x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

# Find frequencies
values, counts = np.unique(x, return_counts=True)

# Find maximum frequency
max_count = counts.max()

# Find all most frequent values
most_frequent_values = values[counts == max_count]

print("Array x:", x)
print("Most frequent value(s):", most_frequent_values)
print("Maximum frequency:", max_count)

for value in most_frequent_values:
    indices = np.where(x == value)[0]
    print("Value", value, "occurs at indices:", indices)


# Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = counts.max()

most_frequent_values = values[counts == max_count]

print("\nArray y:", y)
print("Most frequent value(s):", most_frequent_values)
print("Maximum frequency:", max_count)

for value in most_frequent_values:
    indices = np.where(y == value)[0]
    print("Value", value, "occurs at indices:", indices)
```

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# a) Access 1st row, 2nd column
# Python uses 0-based indexing: row 0, column 1
print("1st row, 2nd column:", arr[0, 1])

# b) Access 3rd row, 1st column
# Python: row 2, column 0
print("3rd row, 1st column:", arr[2, 0])
```

```python
import numpy as np

# Create 1-D array named Lakshay
Lakshay = np.linspace(10, 100, 25)

print("Array:")
print(Lakshay)

# Dimensions
print("\nDimensions:", Lakshay.ndim)

# Shape
print("Shape:", Lakshay.shape)

# Total number of elements
print("Total elements:", Lakshay.size)

# Data type
print("Data type:", Lakshay.dtype)

# Total number of bytes
print("Total bytes consumed:", Lakshay.nbytes)

# Transpose using reshape()
Lakshay_transpose = Lakshay.reshape(1, 25)

print("\nTranspose using reshape():")
print(Lakshay_transpose)

print("Shape after reshape:", Lakshay_transpose.shape)

# Transpose using T attribute
Lakshay_T = Lakshay.T

print("\nUsing T attribute:")
print(Lakshay_T)

print("Shape using T:", Lakshay_T.shape)

print("\nCan we do the same with T attribute?")
print("For a 1-D NumPy array, T does not change its shape.")
print("Therefore, reshape() is needed to convert it into a row/column 2-D array.")
```

```python
import numpy as np

# Create 3 x 4 array
ucs420_Lakshay = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Lakshay)

# Mean
print("\nMean:", np.mean(ucs420_Lakshay))

# Median
print("Median:", np.median(ucs420_Lakshay))

# Maximum
print("Maximum:", np.max(ucs420_Lakshay))

# Minimum
print("Minimum:", np.min(ucs420_Lakshay))

# Unique elements
print("Unique elements:", np.unique(ucs420_Lakshay))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Lakshay = ucs420_Lakshay.reshape(4, 3)

print("\nReshaped array (4 x 3):")
print(reshaped_ucs420_Lakshay)

# Resize to 2 rows and 3 columns
resized_ucs420_Lakshay = np.resize(ucs420_Lakshay, (2, 3))

print("\nResized array (2 x 3):")
print(resized_ucs420_Lakshay)
```
